# End-to-End Reproduction Driver

This notebook mirrors the command-line evaluator path. For formal artifact evaluation, `make reproduce` and `make verify` remain the canonical commands.

If Homebrew Python 3.11 is installed, the notebook will try the Homebrew executable directly. If `python3.11 -m venv` fails with an `ensurepip` or `pyexpat` error, reinstall the formula first:
`brew reinstall expat python@3.11`

## 1. Run the complete reproduction pipeline

```python
import shutil
import subprocess, sys
from pathlib import Path

# Resolve repository root even when Jupyter starts in a different working directory.
ROOT = Path.cwd()
for candidate in [ROOT, *ROOT.parents]:
    if (candidate / "RQ2_Prompt_Effectiveness_Modeling").exists() and (candidate / "Dataset_Construction").exists():
        ROOT = candidate
        break

PY311 = shutil.which("python3.11")
if not PY311:
    homebrew_python = Path("/opt/homebrew/opt/python@3.11/bin/python3.11")
    if homebrew_python.exists():
        PY311 = str(homebrew_python)

if not PY311:
    raise RuntimeError(
        "python3.11 was not found on PATH or at /opt/homebrew/opt/python@3.11/bin/python3.11. "
        "Install Python 3.11 and create a 3.11 venv before running this notebook."
    )

subprocess.check_call(["make", f"PYTHON={PY311}", "reproduce"], cwd=ROOT)
```

## 2. Verify expected outputs

```python
if not PY311:
    raise RuntimeError(
        "python3.11 was not found on PATH or at /opt/homebrew/opt/python@3.11/bin/python3.11. "
        "Install Python 3.11 and create a 3.11 venv before running verification."
    )

subprocess.check_call(["make", f"PYTHON={PY311}", "verify"], cwd=ROOT)
```
