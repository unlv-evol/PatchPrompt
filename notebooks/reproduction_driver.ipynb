# End-to-End Reproduction Driver

This notebook mirrors the command-line evaluator path. For formal artifact evaluation, `make reproduce` and `make verify` remain the canonical commands.

## 1. Run the complete reproduction pipeline

```python
import subprocess, sys
from pathlib import Path
ROOT = Path.cwd()
if not (ROOT / 'Makefile').exists():
    ROOT = ROOT.parent
subprocess.check_call(['make','reproduce'], cwd=ROOT)
```

## 2. Verify expected outputs

```python
subprocess.check_call(['make','verify'], cwd=ROOT)
```

## 3. Review generated output manifests

```python
import pandas as pd
for name in ['tables','figures','diagnostics']:
    path = ROOT / 'results' / 'manifests' / f'{name}_manifest.csv'
    print('\n', path)
    display(pd.read_csv(path).head())
```
